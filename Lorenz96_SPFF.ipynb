#### 1. The deterministic Particle Flow Filter (PFF)

The PFF iteratively pushes forward a set of interacting particles from the prior $p(\mathbf{x})$ to samples from the posterior $p(\mathbf{x}|\mathbf{y})$, where $\mathbf{x} \in \mathbb{R}^{n_x}$ is the model state vector and $\mathbf{y} \in \mathbb{R}^{n_y}$ represents observations, without any reweighting or resampling strategies. The PFF updates an ensemble of $N_p$ particles $\lbrace\mathbf{x}^i \rbrace^{N_p}_{i=1}$ via a gradient velocity field $\mathbf{f}: \mathbb{R}^{n_x}\rightarrow\mathbb{R}^{n_x}$ and follows the dynamics described by a set of ordinary coupled differential equations evolving over pseudo-time $s$. In practice, the PFF is solved in discrete dynamics for each particle:

$$
    \mathbf{x}^i_{s+1}=\mathbf{x}^i_{s}+\Delta s\mathbf{f}(\mathbf{x}^i_{s}),
$$

where $\Delta s$ is the pseudo-time step and $\mathbf{x}^i_{s}$ is the $i^{th}$ particle at pseudo time $s$. The velocity field $\mathbf{f}$ for the $i^{th}$ particle at pseudo time $s$ is:

$$
    \mathbf{f}(\mathbf{x}_s^i) = \frac{1}{N_p} \mathbf{D} \sum_{j=1}^{N_p} \left[ \mathbf{K}(\mathbf{x}_s^{i}, \mathbf{x}_s^{j}) \nabla_{\mathbf{x}_s^j} \log p(\mathbf{x}_s^{j} | \mathbf{y}) + \nabla_{\mathbf{x}_s^j} \cdot \mathbf{K}(\mathbf{x}_s^{i}, \mathbf{x}_s^{j}) \right],
$$

where $\mathbf{D}$ is a positive-definite matrix. Here, we choose $\mathbf{D}$ as the prior covariance $\mathbf{B}$ and can be viewed as a preconditioning matrix, and $\mathbf{K}$ is a kernel matrix of size $n_x\times n_x$. Here, the kernel $\mathbf{K}$ characterizes the distance between any two particles.

In the deterministic PFF, a matrix-valued kernel proposed by Hu and van Leeuwen (2021) is used and is defined as:

$$
    \mathbf{K}(\mathbf{x}^i, \mathbf{x}^j) = \text{diag}([K_{(1)}(\mathbf{x}^i, \mathbf{x}^j), K_{(2)}(\mathbf{x}^i, \mathbf{x}^j), \ldots, K_{(n_x)}(\mathbf{x}^i, \mathbf{x}^j)])
$$

with components:

$$
    K_{(a)}(\mathbf{x}^i, \mathbf{x}^j) = K_{(a)}(x^i_{(a)}, x^j_{(a)}) = \exp\left(-\frac{1}{2}\frac{(x^i_{(a)} - x^j_{(a)})^2}{\alpha\sigma_{(a)}^2}\right)
$$

where $\sigma_{(a)}$ denotes the standard deviation of the $a$-th component for the prior. For the matrix-valued kernel, $\alpha$ is chosen to be the reciprocal of the number of ensemble members.

We assume Gaussian prior and observation errors (though they are not necessary to be Gaussian), so the gradient of the logarithm of the posterior can be expressed as

$$
    \nabla_{\mathbf{x}} \log p(\mathbf{x}|\mathbf{y}) = \mathbf{H}^T \mathbf{R}^{-1} (\mathbf{y} - H(\mathbf{x})) -\mathbf{B}^{-1} (\mathbf{x} - \mathbf{x_b})
$$

Here, $H$ is an observation operator that maps the state space to the observational space, and $\mathbf{H}$ denotes the linearization of $H$ at $\mathbf{x}$. The $\mathbf{R}$ is the observation error covariance matrix, and $\mathbf{x_b}$ is the prior mean.

The first term in the Equation acts as an attracting force that pushes the particles toward the mode of the posterior. The second term represents a repulsive force, pushing particles apart and creating separation within the kernel scale.

Note that the PFF operates similarly to an ensemble of 3D-Vars method, but particles can interact with each other through the kernel matrix during updates. When $N_p=1$, the particle flow reduced to standard variational methods.

#### 2. The Stochastic Particle Flow Filter (SPFF)

The SPFF incorporates an additional Gaussian noise term into the particle flow. The update of $i^{th}$ particle at pseudo time $s$ for SPFF can be written as:

$$
    \mathbf{x}^i_{s+1}=\mathbf{x}^i_{s}+\Delta s \Bigl\lbrace \frac{1}{N_p} \mathbf{D} \sum_{j=1}^{N_p} \left[ \mathbf{K}(\mathbf{x}_s^{i}, \mathbf{x}_s^{j}) \nabla_{\mathbf{x}_s^j} \log p(\mathbf{x}_s^{j} | \mathbf{y}) + \nabla_{\mathbf{x}_s^j} \cdot \mathbf{K}(\mathbf{x}_s^{i}, \mathbf{x}_s^{j}) \right] \Bigr\rbrace + \sqrt{2\Delta s}\boldsymbol{\beta}^i
$$

The deterministic term in the brace is the same gradient velocity field as in PFF, but uses a different kernel matrix. The kernel matrix for SPFF is selected to be a diagonal and isotropic scalar Gaussian function $\mathbf{K}(\mathbf{x}_s^{i}, \mathbf{x}_s^{j}) = K(\mathbf{x}_s^{i}, \mathbf{x}_s^{j})\mathbf{I_{n_x}}$, where $\mathbf{I_{n_x}}$ is an identity matrix of size $n_x\times n_x$, and

$$
    K(\mathbf{x}_s^i, \mathbf{x}_s^{j}) = \exp \left( -\frac{1}{2} (\mathbf{x}_s^i - \mathbf{x}_s^{j})^T (\alpha\mathbf{B})^{-1} (\mathbf{x}_s^i - \mathbf{x}_s^{j}) \right)
$$

where $\alpha$ is an adjustable parameter defined as the kernel width. We set $\alpha=1000$ as the default value for the scalar kernel.

In the noise term, $\boldsymbol{\beta}^i$ represents the noise associated with the $i^{th}$ particle

$$
    \boldsymbol{\beta}^i=(\mathbf{L}_{\mathbf{D}}\mathbf{N}_{{n_x}\times{N_p}}\mathbf{L}_{\overline{\mathbf{k}}}^T)_i
$$

and the subscript $i$ on the right-hand side denotes $i^{th}$ column vector of the matrix. Here, $\mathbf{L}_{\mathbf{D}}$ and $\mathbf{L}_{\overline{\mathbf{k}}}$ are the lower triangular Cholesky decompositions of the localized prior covariance matrix $\mathbf{D}$, and the kernel Gram matrix $\overline{\mathbf{k}} \in \mathbb{R}^{{N_p}\times{N_p}}$, respectively, while  $\mathbf{N}_{{n_x}\times{N_p}}$ is an $n_x \times N_p$ matrix containing standard Gaussian noise generated from $\mathcal{N}(0,\,\mathbf{I}_{{N_p}{n_x}\times{N_p}{n_x}})$. The kernel Gram matrix $\overline{\mathbf{k}}$ is defined by

$$
    \overline{\mathbf{k}} = \frac{1}{N_p}
    \begin{bmatrix}
        K(\mathbf{x}_s^{1}, \mathbf{x}_s^{1}) & K(\mathbf{x}_s^{1}, \mathbf{x}_s^{2}) & \cdots & K(\mathbf{x}_s^{1}, \mathbf{x}_s^{N_p}) \\
        K(\mathbf{x}_s^{2}, \mathbf{x}_s^{1}) & K(\mathbf{x}_s^{2}, \mathbf{x}_s^{2}) & \cdots & K(\mathbf{x}_s^{2}, \mathbf{x}_s^{N_p}) \\
        \vdots  & \vdots  & \ddots & \vdots  \\
        K(\mathbf{x}_s^{N_p}, \mathbf{x}_s^{1}) & K(\mathbf{x}_s^{N_p}, \mathbf{x}_s^{2}) & \cdots & K(\mathbf{x}_s^{N_p}, \mathbf{x}_s^{N_p})
    \end{bmatrix}
$$

This stochastic variant acts as a stochastic gradient Markov Chain Monte Carlo sampler that is unbiased for any ensemble size and asymptotically converges to the exact posterior distribution.

#### 3. Adaptive kernel width for the nonlinear observation operator

The kernel width $\alpha$ controls how far apart two particles can be while still interacting with each other. For the nonlinear observation operator $H(\mathbf{x})=\mathbf{x}^2$, we can choose to rescale it at every assimilation cycle by a factor $\rho$ diagnosed from the prior ensemble:

$$
    \rho_{(a)} = \left( \frac{\sigma(|x_{(a)}|)}{\sigma(x_{(a)})} \right)^2, \qquad \alpha = \alpha_0 \frac{1}{n_x}\sum_{a=1}^{n_x} \rho_{(a)}
$$

where $\sigma$ denotes the standard deviation over the $N_p$ prior particles, $x_{(a)}$ is the $a$-th component of the state, and $\alpha_0$ is the fixed (base) kernel width. Since $|\overline{x_{(a)}}| \leq \overline{|x_{(a)}|}$, we always have $\rho_{(a)} \in (0,1]$, so the adaptive kernel width can only narrow the kernel:

The factor is diagnosed once per assimilation cycle from the prior ensemble, so $\alpha$ follows the sign ambiguity of the ensemble in time. This option is designed for the nonlinear observation operator with the scalar kernel of the SPFF, but since it only rescales $\alpha_0$, it can be applied to the matrix-valued kernel of the PFF as well.

### 1. Set up the Lorenz 96 model

$\frac{dx_{(a)}}{dt} = (x_{(a+1)} - x_{(a-2)})x_{(a-1)} - x_{(a)} + F$  
  
$x_{(a)}$: the a-th component of model state $x$  
  
$F$: the external forcing, set to a common value of 8  

Here, we set the number of state variables $x$ to be 1000 (can be smaller, e.g., 100, for testing with less computational cost).

```python
# Lorenz 96 model parameters
Nx = 1000  # Number of components
F = 8.0 * np.ones(Nx)  # Forcing term

# Function defining the Lorenz 96 model equations
@jit(nopython=True)
def lorenz_96(x, F):
    dxdt = np.zeros_like(x)
    for i in range(Nx):
        dxdt[i] = (x[(i + 1) % Nx] - x[i - 2]) * x[i - 1] - x[i] + F[i]
    return dxdt

# Fourth-order Runge-Kutta integration
@jit(nopython=True)
def runge_kutta4(x, F, dt):
    k1 = lorenz_96(x, F)
    k2 = lorenz_96(x + 0.5 * dt * k1, F)
    k3 = lorenz_96(x + 0.5 * dt * k2, F)
    k4 = lorenz_96(x + dt * k3, F)

    x_new = x + (dt / 6.0) * (k1 + 2 * k2 + 2 * k3 + k4)
    return x_new
```

### 2. Set up the perfect model  experiments
#### 2.1 Nature run (true state)
Integrate the model for 2500 time steps. One model time step is set to be 0.01.

```python
# Time parameters
timesteps = 2500
dt = 0.01

# Initial conditions
np.random.seed(0)
x0 = np.random.rand(Nx)

# Time integration
x_t = np.zeros((Nx, timesteps))
x_t[:, 0] = x0
for t in range(1, timesteps):
    if t == 1:
        for i in range(Nx):
            if np.mod(i+1, 5) == 0:
                F[i] = 9.0
            else:
                F[i] = 8.0
    else:
        F = 8.0 * np.ones(Nx)
    x_t[:, t] = runge_kutta4(x_t[:, t-1], F, dt)
```

### 3. The Deterministic Particle Flow Filter (PFF) and Stochastic Particle Flow Filter (SPFF)
#### 3.1 The PFF and SPFF code

```python
# The Deterministic Particle Flow Filter
def PFF(prior_ens, y, obs_eps, ensemble_size, iter_num, alpha, ds):

    # Input to the algorithm
    R_matrix = np.eye(Ny) * obs_eps  #(Ny,Ny)
    R_matrix_inv = np.linalg.inv(R_matrix)

    # Assume a Gaussian prior in this pseudo code
    prior_mean = np.mean(prior_ens, axis=1)  #(Nx,)

    # generate ensemble perturbation matrix
    ens_perturb = prior_ens - np.tile(prior_mean, (ensemble_size,1)).T  #(Nx,ensemble_size)

    # generate ensemble covariance matrix
    prior_cov = np.matmul(ens_perturb, ens_perturb.T) / (ensemble_size - 1)  #(Nx,Nx)

    # localization

    # generate a distance matrix
    r_in = 4    #decorrelation length scale
    i, j = np.meshgrid(np.arange(Nx), np.arange(Nx), indexing='ij')
    dist_matrix = np.exp(-np.square(np.abs(i - j)/r_in))

    prior_cov = prior_cov*dist_matrix  #(Nx,Nx)
    prior_cov_inv = np.linalg.inv(prior_cov) #(Nx,Nx)

    xs_r = np.zeros((Nx, ensemble_size, iter_num+1))
    xs = prior_ens.copy()  #(Nx,ensemble_size)
    xs_r[:,:,0] = xs.copy()

    # iter_norm_less = 0
    iter_norm_large = 0
    norm_2 = np.zeros((iter_num+1))
    s = 0
    # if s < iter_num:
    for iter in range(1000):
        gradient_log_post = np.zeros((Nx, ensemble_size))
        for i in range(ensemble_size):
            # modeled observation

            if obs_type == 'linear':
                yb = xs[::4, i]
                H = np.zeros((Ny, Nx))
                # linearized observation operator
                for j in range(Ny):
                    H[j, 4*j] = 1
            elif obs_type == 'square':
                yb = np.square(xs[::4, i])
                H = np.zeros((Ny, Nx))
                # linearized observation operator
                for j in range(Ny):
                    H[j, 4*j] = 2*xs[::4, i][j]

            # gradient of log posterior
            gradient_log_post[:, i] = np.matmul(np.matmul(H.T, R_matrix_inv), (y - yb))- np.matmul(prior_cov_inv, (xs[:, i]-prior_mean)) #(Nx,)

        I = np.zeros((Nx, ensemble_size))

        d, i, j = np.meshgrid(np.arange(Nx), np.arange(ensemble_size), np.arange(ensemble_size), indexing='ij')
        K = np.exp(-0.5*np.square((xs[d, i]-xs[d, j]))/(alpha*prior_cov[d, d]))
        dK = (xs[d, i]-xs[d, j])/(alpha*prior_cov[d, d])*K
        I = np.sum((K*gradient_log_post[d, j]+dK)/ensemble_size, axis=-1)

        norm_2[s] = np.linalg.norm(I)
        fs = np.matmul(prior_cov, I)
        # print(fs)
        # print(norm_2[s], norm_2[s-1])

        # Apply the adaptive step size (learning rate)

        if s == 0:
            xs_r[:,:,s+1] = xs_r[:,:,s] + ds*fs
            s = s+1
            iter_norm_large += 1
            xs = xs_r[:,:,s].copy()
        elif s >=1 and norm_2[s] > 1.02*norm_2[s-1]:
            ds = ds/1.5
            s = s-1
            xs = xs_r[:,:,s].copy()
            iter_norm_large = 0
            # print('[Note] The eps has been changed to ', ds, '... Redo the PFF based on this eps')
        elif iter_norm_large >= 7 and norm_2[s] <= 1.02*norm_2[s-1]:
            ds = ds*1.5
            xs_r[:,:,s+1] = xs_r[:,:,s] + ds*fs
            s = s+1
            iter_norm_large = 0
            xs = xs_r[:,:,s].copy()
        else:
            xs_r[:,:,s+1] = xs_r[:,:,s] + ds*fs
            s = s+1
            iter_norm_large += 1
            xs = xs_r[:,:,s].copy()

        # print('iter = ', iter)
        # xs_r[:,:,iter+1] = xs.copy()
        # print('iter = ', s)
        # print('ds = ', ds)

        # print(np.max(np.abs(xs)))
        if s >= iter_num:
            # print('The PFF has been terminated after ', s, ' iterations')
            break
    return xs , xs_r

def SPFF(prior_ens, y, obs_eps, ensemble_size, iter_num, alpha, ds, assim_period, integ_time):

    # Input to the algorithm
    R_matrix = np.eye(Ny) * obs_eps  #(Ny,Ny)
    R_matrix_inv = np.linalg.inv(R_matrix)

    # Assume a Gaussian prior in this pseudo code
    prior_mean = np.mean(prior_ens, axis=1)  #(Nx,)

    # generate ensemble perturbation matrix
    ens_perturb = prior_ens - np.tile(prior_mean, (ensemble_size,1)).T  #(Nx,ensemble_size)

    # generate ensemble covariance matrix
    prior_cov = np.matmul(ens_perturb, ens_perturb.T) / (ensemble_size - 1)  #(Nx,Nx)

    # localization

    # generate a distance matrix
    r_in = 4    #decorrelation length scale
    i, j = np.meshgrid(np.arange(Nx), np.arange(Nx), indexing='ij')
    dist_matrix = np.exp(-np.square(np.abs(i - j)/r_in))

    prior_cov = prior_cov*dist_matrix  #(Nx,Nx)
    prior_cov_inv = np.linalg.inv(prior_cov)

    L_prior_cov = scipy.linalg.cholesky(prior_cov, lower=True)

    xs_r = np.zeros((Nx, ensemble_size, iter_num+1))
    xs = prior_ens.copy()  #(Nx,ensemble_size)
    xs_r[:,:,0] = xs.copy()

    # iter_norm_less = 0
    iter_norm_large = 0
    norm_2 = np.zeros((iter_num+1))
    # norm_2[0] = 10e8
    s = 0
    # if s < iter_num:
    for iter in range(1000):
        gradient_log_post = np.zeros((Nx, ensemble_size))
        for i in range(ensemble_size):
            # modeled observation
            if obs_type == 'linear':
                yb = xs[::4, i]
                H = np.zeros((Ny, Nx))
                # linearized observation operator
                for j in range(Ny):
                    H[j, 4*j] = 1
            elif obs_type == 'square':
                yb = np.square(xs[::4, i])
                H = np.zeros((Ny, Nx))
                # linearized observation operator
                for j in range(Ny):
                    H[j, 4*j] = 2*xs[::4, i][j]

            # gradient of log posterior
            gradient_log_post[:, i] = np.matmul(np.matmul(H.T, R_matrix_inv), (y - yb))- np.matmul(prior_cov_inv, (xs[:, i]-prior_mean)) #(Nx,)

        I = np.zeros((Nx, ensemble_size))

        K = np.zeros((ensemble_size, ensemble_size))
        A = (1.0/alpha)*prior_cov_inv
        A_tran = A.T
        for i in range(ensemble_size):
            for j in range(ensemble_size):
                diff_xs = xs[:, i]-xs[:, j]
                N_diff_xs = np.matmul(A,diff_xs)
                K[i,j] = np.exp(-0.5*np.matmul(diff_xs.T, N_diff_xs))
                dK = np.matmul(A_tran, diff_xs)*K[i,j]
                I[:, i] = I[:, i] + (K[i,j]*gradient_log_post[:, j]+dK)/ensemble_size

        # Calculate the stochastic part

        # Generate Gaussian random numbers with mean 0 and variance 1
        # set the seed
        # np.random.seed(4)
        np.random.seed(4+s+iter_num*(integ_time//assim_period-1))
        random_numbers_stochastic = np.random.normal(0, 1, Nx*ensemble_size)

        # Calculate the lower triangular Cholesky decomposition of K matrix
        LK = scipy.linalg.cholesky(K/ensemble_size, lower=True)

        # Efficient calculation of noise, replacing the L_full approach",
        R_Nx_ens = random_numbers_stochastic.reshape(Nx, ensemble_size)
        noise = compute_noise_fast(L_prior_cov, R_Nx_ens, LK.T, np.sqrt(2))

        # print(I.shape)

        norm_2[s] = np.linalg.norm(I)

        # if norm_2[s] < stop_cri_percentage * norm_2[0]:
        #     print('The PFF has been terminated after ', s, ' iterations')
        #     break

        fs = np.matmul(prior_cov, I)
        # print(fs)
        print(norm_2[s], norm_2[s-1])

        # Apply the adaptive step size (learning rate)

        if s == 0:
            # xs_r[:,:,s+1] = xs_r[:,:,s] + ds*fs
            xs_r[:,:,s+1] = xs_r[:,:,s] + ds*fs + np.sqrt(ds)*noise
            s = s+1
            iter_norm_large += 1
            xs = xs_r[:,:,s].copy()
        elif s >= 1 and norm_2[s] > 1.02*norm_2[s-1]:
            ds = ds/1.5
            s = s-1
            xs = xs_r[:,:,s].copy()
            iter_norm_large = 0
            # print('[Note] The eps has been changed to ', ds, '... Redo the PFF based on this eps')
        elif iter_norm_large >= 7 and norm_2[s] <= 1.02*norm_2[s-1]:
            ds = ds*1.5
            # xs_r[:,:,s+1] = xs_r[:,:,s] + ds*fs
            xs_r[:,:,s+1] = xs_r[:,:,s] + ds*fs + np.sqrt(ds)*noise
            s = s+1
            iter_norm_large = 0
            xs = xs_r[:,:,s].copy()
        else:
            # xs_r[:,:,s+1] = xs_r[:,:,s] + ds*fs
            xs_r[:,:,s+1] = xs_r[:,:,s] + ds*fs + np.sqrt(ds)*noise
            s = s+1
            iter_norm_large += 1
            xs = xs_r[:,:,s].copy()

        # print('iter = ', iter)
        # xs_r[:,:,iter+1] = xs.copy()
        print('iter = ', s)
        # print('ds = ', ds)

        print(np.max(np.abs(xs)))
        if s >= iter_num:
            print('The PFF has been terminated after ', s, ' iterations')
            break
    return xs , xs_r

@jit(nopython=True)
def compute_noise_fast(L_prior_cov, random_numbers, LK_T, sqrt_2):
    """Fast noise computation using Numba JIT"""
    temp = np.dot(L_prior_cov, random_numbers)
    return sqrt_2 * np.dot(temp, LK_T)
```

#### 3.2 Running the sequential DA experiment with PFF or SPFF (observations are assimilated every 20 model time step)
1. Set the number of iterations
2. Choose PFF or SPFF
3. Choose a fixed or an adaptive kernel width

```python
iter_num = 500      #number of iterations (150 for linear, 500 for square)
Assim_method = 'SPFF'  # Please type ['PFF', 'SPFF']
kernel_width_type = 'adaptive'  # Please type ['fixed', 'adaptive']. The adaptive is only for SPFF
```

Run the model

```python
# Fixed parameters for model and (S)PFF
integration_time_step = 500   # total number of time steps for sequential Data Assimilation (maximum: 1500)
F = 8.0 * np.ones(Nx)
if Assim_method == 'PFF':
    alpha_0 = 1/ensemble_size    # the base width of the kernel
elif Assim_method == 'SPFF':
    alpha_0 = Nx    # the base width of the kernel

if obs_type == 'linear':
    ds = 0.05           # step size
elif obs_type == 'square':
    ds = 0.001

# Number of assimilation cycles
Number_of_assim_cycles = int(integration_time_step/assim_period)

# Initial conditions
prior_ens_PFF = np.zeros((Nx, ensemble_size, int(integration_time_step/assim_period)+1))
posterior_ens_PFF = np.zeros((Nx, ensemble_size, int(integration_time_step/assim_period)+1))
x_PFF = np.zeros((Nx, ensemble_size, integration_time_step+1))

posterior_ens_PFF[:,:,0] = prior_ens_0.copy()  #(Nx,ensemble_size)
x_PFF[:,:,0] = prior_ens_0.copy()  #(Nx,ensemble_size)

# Set the kernel width and keep a record of the width used at each assimilation cycle
alpha = alpha_0
alpha_list = np.zeros((Number_of_assim_cycles+1))
alpha_list[0] = alpha_0
print("The", kernel_width_type, "kernel width is used, and the base kernel width is", alpha_0)

# Sequential DA for PFF with 20 time-steps frequency
for t in range(1, integration_time_step+1):
# for t in range(1, integration_time_step):

    # Model integration
    for i in range(ensemble_size):
        x_PFF[:, i, t] = runge_kutta4(x_PFF[:, i, t-1], F, dt)

    if np.mod(t, assim_period) == 0:
        print("Implement PFF at time step ", t, " ...")

        # set the prior ensemble at the assimilation time
        prior_ens_PFF[:, :, int(t/assim_period)] = x_PFF[:, :, t]

        # set the kernel width for this assimilation cycle
        if kernel_width_type == 'fixed':
            alpha = alpha_0
        elif kernel_width_type == 'adaptive':
            # adjust the base kernel width with a factor rho diagnosed from the prior ensemble
            prior_ens_abs = np.abs(prior_ens_PFF[:, :, int(t/assim_period)])  #(Nx,ensemble_size)
            rho = np.square(np.std(prior_ens_abs, axis=1)/np.std(prior_ens_PFF[:, :, int(t/assim_period)], axis=1))  #(Nx,)
            alpha = alpha_0*np.mean(rho)
            print("The kernel width is ", alpha, " at time step ", t, " ...", "The mean of rho is ", np.mean(rho))
        alpha_list[int(t/assim_period)] = alpha

        # run the PFF algorithm
        if Assim_method == 'PFF':
            [posterior_ens_PFF[:,:,int(t/assim_period)], xs_r] = PFF(prior_ens_PFF[:, :, int(t/assim_period)], obs_set[:, int(t/assim_period)], obs_eps, ensemble_size, iter_num, alpha, ds)
        elif Assim_method == 'SPFF':
            [posterior_ens_PFF[:,:,int(t/assim_period)], xs_r] = SPFF(prior_ens_PFF[:, :, int(t/assim_period)], obs_set[:, int(t/assim_period)], obs_eps, ensemble_size, iter_num, alpha, ds, assim_period, t)

        # update the model state by posterior ensemble
        x_PFF[:, :, t] = posterior_ens_PFF[:,:,int(t/assim_period)]

        # save the ensemble trajectory during the first assimilation period
        if t == assim_period*1:
            xs_r1 = xs_r.copy()
```
